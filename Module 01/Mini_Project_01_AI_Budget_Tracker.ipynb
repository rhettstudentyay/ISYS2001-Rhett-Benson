## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
Write a Python program that:
- Asks the user for their monthly income.
- Asks the user for up to three separate expenses (e.g., rent, groceries, utilities).
- Calculates the total expenses.
- Calculates the remaining budget (income - total expenses).
- Prints a summary of the income, expenses, and remaining budget.
- If the remaining budget is negative, print a clear warning message indicating that the user is over budget and by how much.
- Ensure robust input handling, using try-except blocks for numerical inputs.
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# [Paste your code here]

def get_numerical_input(prompt_message):
    while True:
        try:
            value = float(input(prompt_message))
            if value < 0:
                print("Please enter a non-negative number.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a numerical value.")

monthly_income = get_numerical_input("Enter your monthly income: $")

expenses = []
# Loop up to 3 times to get expenses
for i in range(3):
    while True:
        add_expense = input(f"Do you want to add expense {i+1}? (yes/no): ").lower().strip()
        if add_expense in ['yes', 'no']:
            break
        else:
            print("Invalid input. Please type 'yes' or 'no'.")

    if add_expense == 'yes':
        expense_name = input(f"Enter name for expense {i+1}: ")
        # Basic validation for expense name
        if not expense_name:
            print("Expense name cannot be empty. Skipping this expense.")
            continue
        expense_amount = get_numerical_input(f"Enter amount for {expense_name}: $")
        expenses.append({'name': expense_name, 'amount': expense_amount})
    elif add_expense == 'no':
        break # Stop adding expenses if the user says 'no'

total_expenses = sum(expense['amount'] for expense in expenses)
remaining_budget = monthly_income - total_expenses

print("\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if expenses:
    for expense in expenses:
        print(f"  - {expense['name']}: ${expense['amount']:.2f}")
else:
    print("  No expenses entered.")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

if remaining_budget < 0:
    print(f"\nWARNING: You are over budget by ${abs(remaining_budget):.2f}! Consider adjusting your expenses.")
else:
    print("\nGreat! You are within your budget.")
```
