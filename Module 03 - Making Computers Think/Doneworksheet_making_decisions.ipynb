````python
def classify_gpa(gpa):
    if gpa >= 3.5:
        return "Great"
    elif gpa >= 3.0:
        return "Good"
    elif gpa >= 2.5:
        return "Probation"
    else:
        return "Academic Warning"

# Test cases based on your predictions
students = {
    "Student A": 3.7,
    "Student B": 2.9,
    "Student C": 3.0,
    "Student D": 2.4, # Added a case for Academic Warning
    "Student E": 1.8  # Added another case for Academic Warning
}

print("=== Academic Standing Classification ===")
for name, gpa in students.items():
    status = classify_gpa(gpa)
    print(f"{name} (GPA: {gpa}): {status}")
````

````python
# Your study recommendation system

# Try these test cases once it works:
# - studied 6 hours, stress level 8
# - studied 1 hour, stress level 3
# - studied 0 hours, exam tomorrow

````

````python
def get_study_recommendation(hours_studied, stress_level):
    if hours_studied >= 3 and stress_level <= 5:
        return "Take a break and review key concepts."
    elif hours_studied < 1 and stress_level >= 7:
        return "Prioritize rest, then start with quick review sessions."
    elif hours_studied >= 2 and stress_level >= 6:
        return "Focus on understanding difficult topics for short bursts, then switch to easier material."
    else:
        return "Continue with your current study plan, and take breaks as needed."

print("=== Study Recommendations ===")

# Test cases
# - studied 6 hours, stress level 8
recommendation_1 = get_study_recommendation(6, 8)
print(f"Studied 6 hours, Stress 8: {recommendation_1}")

# - studied 1 hour, stress level 3
recommendation_2 = get_study_recommendation(1, 3)
print(f"Studied 1 hour, Stress 3: {recommendation_2}")

# - studied 0 hours, exam tomorrow (Assuming 'exam tomorrow' implies high stress, e.g., 9)
# Note: "exam tomorrow" is not a direct input factor in your rules, so interpreting it as high stress.
recommendation_3 = get_study_recommendation(0, 9)
print(f"Studied 0 hours, Stress 9 (Exam Tomorrow): {recommendation_3}")

# Additional test case to cover another rule
recommendation_4 = get_study_recommendation(2.5, 6.5)
print(f"Studied 2.5 hours, Stress 6.5: {recommendation_4}")
````

### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: GPA
Factor 2: Test Score
Factor 3: Essay
Decision logic: If gpa is under a certian level then admission is failed. If test score is too low then admission is failed. If essay is bad then admission is failed.
```

````python
# Your admissions decision system
# Tip: for "at least 2 of 3 conditions true", try counting how many are True.

def evaluate_admission(gpa, test_score, essay_score):
    # Define thresholds for each factor
    MIN_GPA = 3.0
    MIN_TEST_SCORE = 80  # Assuming out of 100
    MIN_ESSAY_SCORE = 7  # Assuming out of 10

    # Evaluate each condition
    gpa_meets_criteria = (gpa >= MIN_GPA)
    test_score_meets_criteria = (test_score >= MIN_TEST_SCORE)
    essay_meets_criteria = (essay_score >= MIN_ESSAY_SCORE)

    # Count how many conditions are met
    conditions_met = 0
    if gpa_meets_criteria:
        conditions_met += 1
    if test_score_meets_criteria:
        conditions_met += 1
    if essay_meets_criteria:
        conditions_met += 1

    # Decision logic: at least 2 of 3 conditions must be true for admission
    if conditions_met >= 2:
        return "Admitted"
    else:
        return "Not Admitted"

# --- Test Cases ---
print("=== Admissions Decisions ===")

# Student 1: Clearly Admitted (all 3 met)
student1_gpa, student1_test, student1_essay = 3.8, 92, 9
status1 = evaluate_admission(student1_gpa, student1_test, student1_essay)
print(f"Student 1 (GPA: {student1_gpa}, Test: {student1_test}, Essay: {student1_essay}): {status1}")

# Student 2: Admitted (2 of 3 met - GPA and Test)
student2_gpa, student2_test, student2_essay = 3.5, 85, 6
status2 = evaluate_admission(student2_gpa, student2_test, student2_essay)
print(f"Student 2 (GPA: {student2_gpa}, Test: {student2_test}, Essay: {student2_essay}): {status2}")

# Student 3: Not Admitted (only 1 of 3 met - GPA)
student3_gpa, student3_test, student3_essay = 3.2, 70, 5
status3 = evaluate_admission(student3_gpa, student3_test, student3_essay)
print(f"Student 3 (GPA: {student3_gpa}, Test: {student3_test}, Essay: {student3_essay}): {status3}")

# Student 4: Not Admitted (0 of 3 met)
student4_gpa, student4_test, student4_essay = 2.5, 60, 4
status4 = evaluate_admission(student4_gpa, student4_test, student4_essay)
print(f"Student 4 (GPA: {student4_gpa}, Test: {student4_test}, Essay: {student4_essay}): {status4}")

# Student 5: Admitted (2 of 3 met - Test and Essay, lower GPA)
student5_gpa, student5_test, student5_essay = 2.9, 90, 8
status5 = evaluate_admission(student5_gpa, student5_test, student5_essay)
print(f"Student 5 (GPA: {student5_gpa}, Test: {student5_test}, Essay: {student5_essay}): {status5}")
````

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
